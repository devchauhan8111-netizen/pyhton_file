```python
# game 
print( " Welcome to quiz world ")
print(" you have only 2 chances for save your life")

score = 0
chances = 2
q = 1
while q <= 7 and chances > 0:
    
  
    if q == 1:
        print(" Q1 . capital of india ?")
        print(" a) Mumbai")
        print(" b) New Delhi")
        print(" c) chennai")
        
        ans = "b"
    elif q == 2:
        print(" Q2 . Largest planet ?")
        print(" a) Earth")
        print(" b) Mars")
        print(" c) Jupiter")
        ans = "c"
    elif q == 3:
        print(" Q3 . water formula ?")
        print(" a) H2O")
        print(" b) CO2")
        print(" c) O2")
        ans = "a"
    elif q == 4:
        print(" Q4 . Sun rises from ?")
        print(" a) West")
        print(" b) East")
        print(" c) North")
        ans = "b"
    elif q == 5:
        print(" Q5 . Keyboard is a ?")
        print(" a) Input device ")
        print(" b) Output device")
        print(" c) Storage device")
        ans = "b"
    elif q == 6:
        print(" Q6 .Html used for ?")
        print(" a) Web pages")
        print(" b) Games")
        print(" c) Music")
        ans = "a" 
    elif q == 7:
        print(" Q7 . Fastest land animal ?")
        print(" a) Tiger")
        print(" b) Lion")
        print(" c) Cheetah")
        ans = "c"
    player = input("enter your answer")  
    if player.lower() ==  ans:        
        print("nice")
        score += 1
        q += 1
    elif player.lower() == "skip":
        q += 1
        
  
    else:
        chances -= 1
        print(" one life is over" )
    if chances == 0:
            print("you are dead")
        
    elif score == 7:
         print("you are safe")
        
```
