```python
# importing major liberaries

import numpy as np
import pandas as pd
```

```python
df = pd.read_excel('insurance_logistic_dataset_600.xlsx')
```

```python
df
import matplotlib.pyplot as plt
import seaborn as sns 
plt.figure(figsize=(10,3))
sns.scatterplot(data=df,x='Age',y='Has_Insurance')
plt.show()
```

```python
# kde plot
# universal analysis
# distribution
plt.figure(figsize=(10,3))
sns.kdeplot(data=df,x='Age',hue='Has_Insurance')
plt.show()
```

```python
# logistic regression

from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
df
# independent and dependent variables

x = df.iloc[:,0:-1]# dataframe
y = df.iloc[:,-1] # series
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
lr = LogisticRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
pd.concat([y_test.reset_index().drop('index',axis=1),pd.Series(y_pred,name='Predicted')],axis=1)
```

```python
# accuracy

from sklearn.metrics import accuracy_score,confusion_matrix
accuracy_score(y_test,y_pred)
```

```python
pd.DataFrame(confusion_matrix(y_test,y_pred))
```

```python
sns.heatmap(pd.DataFrame(confusion_matrix(y_test,y_pred)),annot=True,cmap='PuBu')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()
```
