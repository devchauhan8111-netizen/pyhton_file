```python
# iloc --> index values
# rows
df.iloc[10:20] # range indexing
df.iloc[[2,3,10,23,45]] # fancy indexing

# columns
df.iloc[:,5:10] # range index
df.iloc[13:20,5:10]
df.iloc[[2,45,56],[4,7,2,5]] #fancy indexing
```
