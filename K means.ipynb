```python
# kmeans
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4,max_iter=1000,random_state=42)
y_means = kmeans.fit_predict(df) # main change!!!!
kmeans
```

```python
pd.Series(y_means).value_counts()
```

```python
cl_means = pd.concat([df,pd.Series(y_means,name='cluster')],axis=1)
```

```python
plt.figure(figsize=(8,4))
sns.scatterplot(data=cl_means,x='cgpa',y='iq',hue='cluster',palette='rainbow')
plt.grid()
plt.show()
```
